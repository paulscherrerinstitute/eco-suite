# eco
The Namespace panel (left) and each opened device are real JupyterLab dock panels via Sidecar -- drag/tab/tile them like any other JupyterLab panel. Click the code cell's collapse arrow on the left to hide it; it's already collapsed by default.

The scope is chosen by the `eco` launcher via the `ECO_SCOPE` environment variable (default `bernina`).

```python
import os
import importlib

scope = os.environ.get("ECO_SCOPE") or "bernina"
lazy = os.environ.get("ECO_LAZY", "1") not in ("0", "false", "False", "")

# Configure lazy initialisation before importing the scope, exactly like the
# shell startup does.
from eco import ecocnf
if lazy:
    ecocnf.startup_lazy = True

mod = importlib.import_module(f"eco.{scope}")
# Each scope module exposes a `namespace` root; fall back to the module itself.
target = getattr(mod, "namespace", mod)

# Sidecar panels (a real Lumino dock widget, JupyterLab-only -- Voila has no
# docking shell for this, see eco.widgets.widget_tray.make_namespace_dashboard
# there instead): the Namespace launcher docks on the left, same Name/
# Required table as eco desktop's dockable panel; each device you open from
# it docks in its own panel on the right. `dashboard` has to stay referenced
# (a notebook-global variable does) or its Sidecars get garbage-collected
# and close.
from eco.widgets.jupyter_sidecar import open_namespace_dashboard
dashboard = open_namespace_dashboard(target)

# Only takes effect here if this notebook is opened with a plain
# "python3" kernel (e.g. bare `eco jupyterlab`, no -s/--scope) --
# with a scope given, eco_cli.py already points this notebook at a
# generated eco-<scope> kernel whose own profile startup does both of
# these (see _register_eco_kernel); install_shell_logger is
# idempotent, so this is harmless, not a double-install, either way.
get_ipython().Completer.use_jedi = False
from eco.widgets import kernel_registry
kernel_registry.install_shell_logger(kind="jupyterlab", label=scope)
```
