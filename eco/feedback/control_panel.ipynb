# eco feedback control panel

Server URL from `ECO_FEEDBACK_URL` (default `http://127.0.0.1:8010`), API key from `ECO_FEEDBACK_API_KEY`. Started by `python -m eco.feedback panel URL`.

```python
import os
from eco.feedback.gui import FeedbackPanel

panel = FeedbackPanel(os.environ.get("ECO_FEEDBACK_URL", "http://127.0.0.1:8010"),
                      api_key=os.environ.get("ECO_FEEDBACK_API_KEY") or None)
panel.widget
```
